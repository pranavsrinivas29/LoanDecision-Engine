```python
import os
import pandas as pd
import numpy as np
import mlflow
```

```python
PROJECT_ROOT = os.path.abspath("..")
DATA_PATH = os.path.join(PROJECT_ROOT, "data", "raw", "loan_approval_dataset.csv")

print("Project root:", PROJECT_ROOT)
print("Data path:", DATA_PATH)
```

```python
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:\n", df.columns.tolist())
print("\nData types:\n", df.dtypes)
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.columns.tolist()
```

```python
df.columns = df.columns.str.strip()
print(df.columns.tolist())
```

```python
target_col = "loan_status"
print(df[target_col].astype(str).str.strip().str.lower().unique())
```

```python
target_col = "loan_status"

df[target_col] = (
    df[target_col]
    .astype(str)
    .str.strip()
    .str.lower()
)

print(df[target_col].value_counts(dropna=False))
```

```python
df.to_csv(os.path.join(PROJECT_ROOT, "data", "raw", "loan_approval_dataset.csv"), index=False)
```

```python
target_col = "loan_status"

print(df[target_col].value_counts(dropna=False))
print(df[target_col].value_counts(normalize=True, dropna=False))
```

```python
mlflow.set_tracking_uri("file:../mlruns")
mlflow.set_experiment("loan_approval_phase1")

print("Tracking URI:", mlflow.get_tracking_uri())

#mlflow.set_experiment("loan_approval_phase1")

with mlflow.start_run(run_name="data_loading_inspection"):
    mlflow.log_param("dataset_path", DATA_PATH)
    mlflow.log_param("num_rows", df.shape[0])
    mlflow.log_param("num_columns", df.shape[1])
    mlflow.log_param("target_column", target_col)

    missing_total = int(df.isnull().sum().sum())
    mlflow.log_metric("total_missing_values", missing_total)

print("Logged to MLflow successfully.")
```
